Missing Values

```python
print("\nMissing Values:")
display(df.isna().sum())
```

Handling missing values

```python
df = df.drop(columns=['Comments'])
```

Checking for the negative values in GHI, DNI, DHI

```python
radiation_cols=['GHI','DNI','DHI']
negative_cols = (df[radiation_cols]<0).sum()
print("Negative values in radiation columns:\n", negative_cols)
```

Checking duplicates

```python
# Exact duplicate rows
dup_count = df.duplicated().sum()
print("Duplicate rows:", dup_count)

# Cardinality (uniqueness) for categoricals
cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
cardinality = {c: df[c].nunique() for c in cat_cols}
print("Cardinality (categoricals):", cardinality)
```

Identify outliers using z-scores

```python
cols_to_check = ['GHI','DNI','DHI', 'ModA','ModB','WS','WSgust']
```

```python
z_scores= np.abs(stats.zscore(df[cols_to_check]))
```

```python
outliers_mask=(z_scores>3).any(axis=1)
```

```python
print(f"\nRows flagged as outliers: {outliers_mask.sum()}")
```

```python
for col in ['GHI','DNI','DHI']:
    df.loc[df[col]<0,col]=0
```

```python
df_clean = df[~outliers_mask].copy()
```

```python
print(f"Original rows: {len(df)}")
print(f"Cleaned rows:{len(df_clean)}")
print(f"Remaining negatives: {(df_clean[['GHI','DNI','DHI']]<0).sum().sum()}")
```

```python
df_clean['Timestamp'] = pd.to_datetime(df_clean['Timestamp'])
```

Univariate aAnalysis 

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
for c in num_cols:
    plt.figure()
    sns.histplot(df[c], kde=True)
    plt.title(f"Distribution: {c}")
    plt.xlabel(c)
    plt.ylabel("Count")
    plt.show()
```

### Univariate Analysis
- The histograms show the distribution of numerical variables. GHI, DNI, and DHI exhibit right-skewed distributions, typical for solar data. Temperature and humidity show more normal distributions, while wind speed is skewed.

Bivariate Analysis

```python
# Bivariate Analysis: Scatter plots between key variables
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
sns.scatterplot(data=df_clean, x='GHI', y='Tamb', alpha=0.5)
plt.title('GHI vs Ambient Temperature')

plt.subplot(2, 2, 2)
sns.scatterplot(data=df_clean, x='GHI', y='RH', alpha=0.5)
plt.title('GHI vs Relative Humidity')

plt.subplot(2, 2, 3)
sns.scatterplot(data=df_clean, x='Tamb', y='RH', alpha=0.5)
plt.title('Temperature vs Humidity')

plt.subplot(2, 2, 4)
sns.scatterplot(data=df_clean, x='WS', y='GHI', alpha=0.5)
plt.title('Wind Speed vs GHI')

plt.tight_layout()
plt.show()
```

### Interpretation Bivariate Analysis 
- Scatter plots reveal relationships between variables:
  - GHI vs Temperature: Shows a positive correlation, indicating higher solar radiation on hotter days.
  - GHI vs Humidity: Negative correlation, as cloudy/humid conditions reduce solar irradiance.
  - Temperature vs Humidity: Inverse relationship, common in weather data.
  - Wind Speed vs GHI: Weak relationship, suggesting wind has limited direct impact on solar radiation.


Plots

## Correlation Matrix & Heatmap (numeric)

```python
#selecting main columns for correlation
corr_cols = ['GHI', 'DNI', 'DHI', 'TModA', 'TModB', 'Tamb', 'RH', 'WS', 'WSgust', 'BP']
correlation_matrix = df_clean[corr_cols].corr()

# 2. Plot the Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='RdBu', center=0)
plt.title('Correlation Heatmap of Solar & Weather Variables')
plt.show()
```

## Interpretation of the Correlation Matrix

The correlation matrix and heatmap provide insights into the linear relationships between the main solar and weather variables:

- **Strong Positive Correlations:**
  - GHI, DNI, and DHI (solar radiation variables) are highly correlated with each other, as expected.
  - Module temperatures (TModA, TModB) and ambient temperature (Tamb) are also strongly correlated, indicating that module temperature closely follows ambient conditions.
- **Negative Correlations:**
  - Relative Humidity (RH) shows a negative correlation with solar radiation variables (GHI, DNI, DHI), meaning that higher humidity is generally associated with lower solar irradiance (cloudier conditions).
  - Wind Speed (WS) and module/ambient temperatures have weak to moderate negative correlations, suggesting that higher wind speeds may help cool the modules and ambient air.
- **Weak or No Correlation:**
  - Barometric Pressure (BP) does not show strong correlation with most other variables, indicating it is relatively independent in this context.

**Key Takeaway:**
- The matrix helps identify which variables are most closely related, which is useful for feature selection, multicollinearity checks, and understanding the physical relationships in the solar dataset.
